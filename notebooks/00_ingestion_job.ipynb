```python
# Paramètre transmis par le Job
dbutils.widgets.text("sample_size", "300000")
sample_size = int(dbutils.widgets.get("sample_size"))

if sample_size <= 0:
    raise ValueError("sample_size doit être strictement positif")

# Préparation des emplacements
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.`5lake_lab`")
spark.sql(
    "CREATE VOLUME IF NOT EXISTS workspace.`5lake_lab`.raw"
)

# Lecture limitée au nombre demandé
df_raw = (
    spark.table("samples.nyctaxi.trips")
    .limit(sample_size)
)

# Export des fichiers consommés par ton pipeline
raw_path = "/Volumes/workspace/5lake_lab/raw/nyc_trips"

(
    df_raw.write
    .mode("overwrite")
    .parquet(raw_path)
)

print("Taille demandée :", sample_size)
print("Lignes exportées :", spark.read.parquet(raw_path).count())
```
